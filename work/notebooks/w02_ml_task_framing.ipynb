## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

This is a scoring / ranking task, not plain classification. The end deliverable isn't a single yes/no label per page — it's a ranked queue, where pages are ordered by a continuous priority score so an editor can work top-N down. Under the hood it's built on a classifier (predicting probability of being a "declining" page), but that probability is the score — it only becomes useful once it's used to rank candidates and evaluated with precision@K, which is what makes it a ranking/scoring problem rather than a classification problem in the strict sense.

*One metric you can defend. What number means 'good'?*

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A hand-written rule (e.g. "flag it if impressions dropped more than X%") only looks at one signal in isolation and picks one threshold for every page, regardless of context. In reality, whether a drop matters depends on how it combines with other signals at once — starting traffic level, how long the decline has run, content age, page type — and those interactions aren't linear or fixed. That's exactly the mess a single if-statement can't capture but a model can learn from data. This isn't a hypothesis on my part — it's demonstrated in this repo's own pipeline: the fixed rule tops out at precision@50 ≈ 0.24, while the trained random forest reaches ≈ 0.74 on the same data. That gap is the evidence that the pattern here needed more than a threshold.
